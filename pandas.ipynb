assigning index

```python
series3 = pd.Series([1,2,3,4], index = ['a','b','c','d'])
print(series3)
```

declaring a datatype

```python
series4 = pd.Series([5,6,7,8], index =['e','f','g','h'], dtype = float)
print(series4)
```

giving a column name

```python
series5 = pd.Series([73,74,76],index = ['suchi','sukanya','barsha'], name= 'roll numbers')
print(series5)
```

for declaring single (scalar) values

```python
scalar_s = pd.Series(0.5)
print(scalar_s)
```

for declaring same value for different index

```python
scalar_s = pd.Series('mca', index =['suchi','sukanya','barsha'])
print(scalar_s)
```

create from dictionary

```python
dict_s = pd.Series({'a':1, 'b':2})
print(dict_s)
```

# DATA FRAMES IN PANDAS

how to create a dataframe

```python
import pandas as pd
```

empty dataframe

```python
empty_df = pd.DataFrame()
print(empty_df)
```

creating dataframe using list

```python
lst = ['sukanya','suchitra','barsha']
print(lst)
```

creating dataframe using the above creates list

```python
df1 = pd.DataFrame(lst)
print(df1)
```

```python
df1
```

```python
lst1 = ['ihcus','aynakus','ahsrab']
print(lst1)
```

```python
df2 = pd.DataFrame(lst1)
print(df2)
```

```python
lst_of_lst =[[1,2,3],[2,3,4],[4,5,6]]
print(lst_of_lst)
```

```python
df2 = pd.DataFrame(lst_of_lst)
print(df2)
```

creating a dictionary

array values should be of same length

```python
dict1 = {'ID':[11,22,33,44],'SN':[1,2,3,4]}
print(dict1)
```

```python
df3 = pd.DataFrame(dict1)
print(df3)
```

creating list of dictionaries

```python
ls_dict = [{'a':1, 'b':2},{'a':3, 'b':4}]
df5 = pd.DataFrame(ls_dict)
df5
```

creating list of dictionaries by using dataframes

```python
ls_dict = [{'a':1, 'b':2},{'a':3, 'b':4, 'c':5}]
df5 = pd.DataFrame(ls_dict)
df5
```

```python
ls_dict = [{'a':3, 'b':4},{'a':6, 'b':7, 'c':8}]
df6 = pd.DataFrame(ls_dict)
df6
```

creating dictionaries of series with the help of dataframes

```python
dict_sr = {'id': pd.Series([1,2,3]), 'sn':pd.Series([111,222,333])}
df6 = pd.DataFrame(dict_sr)
df6
```

```python
dict_ser = {'id': pd.Series([1,2,3]), 'sn':pd.Series([101,202,303])}
df7 = pd.DataFrame(dict_ser)
df7
```
